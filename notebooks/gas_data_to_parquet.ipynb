```python
from pathlib import Path

import polars as pl

data_dir = Path("../data")

for market_dir in (data_dir / "ttf futures", data_dir / "ttf spot"):
    market_name = market_dir.name.replace(" ", "_")

    for data_kind in ("Settlement", "Statistics"):
        source_paths = sorted(market_dir.glob(f"{data_kind}*.xlsx"))
        if not source_paths:
            raise FileNotFoundError(
                f"No {data_kind} workbooks found in {market_dir}"
            )

        frames = [pl.read_excel(path) for path in source_paths]
        merged = pl.concat(frames, how="vertical_relaxed", rechunk=True)

        output_dir = market_dir / data_kind.lower()
        output_dir.mkdir(parents=True, exist_ok=True)
        output_path = output_dir / f"{market_name}.parquet"
        merged.write_parquet(output_path)

        print(
            f"Wrote {merged.height:,} rows and {merged.width} columns to "
            f"{output_path}"
        )
```
